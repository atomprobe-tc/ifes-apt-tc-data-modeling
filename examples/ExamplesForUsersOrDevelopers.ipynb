## Stuttgart TAP-type system files

```python
prefix = f"{os.getcwd()}/../tests/data/stuttgart/examples_with_provenance"
fnm = [
    "apyt_W_calibration_tap_01_trimmed.raw",
    "db.yaml",
    "W_calibration_tap_01_trimmed.txt",
    # "W_calibration_tap_01_trimmed_mass_spectrum_fit.txt",
    # "W_calibration_tap_01_trimmed_mass_spectrum_raw.txt",
    "W_calibration_tap_01_trimmed_xyz.txt",
]
```

```python
from ifes_apt_tc_data_modeling.stuttgart.apyt_reader import (
    ReadStuttgartApytMetadataFileFormat,
    # ReadStuttgartApytSpectrumFitFileFormat,
    ReadStuttgartApytReconstructionFileFormat,
    ReadStuttgartApytSpectrumAlignFileFormat,
)
from ifes_apt_tc_data_modeling.stuttgart.raw_reader import (
    ReadStuttgartApytRawFileFormat,
)

for fpath in fnm:
    if fpath.lower().endswith(".raw"):
        print(fpath)
        stuttgart = ReadStuttgartApytRawFileFormat(f"{prefix}/{fpath}")
        print(stuttgart.number_of_events)
        print(stuttgart.get_raw_time_of_flight())
    elif fpath.lower().endswith("db.yaml"):
        print(fpath)
        stuttgart = ReadStuttgartApytMetadataFileFormat(f"{prefix}/{fpath}")
        print(stuttgart.get_metadata(database_entry=1))
    elif fpath.lower().endswith("_trimmed.txt"):
        print(fpath)
        stuttgart = ReadStuttgartApytSpectrumAlignFileFormat(f"{prefix}/{fpath}")
        print(stuttgart.get_complete_spectrum())
    # TODO add SpectrumFit
    elif fpath.lower().endswith("_trimmed_xyz.txt"):
        print(fpath)
        stuttgart = ReadStuttgartApytReconstructionFileFormat(f"{prefix}/{fpath}")
        print(stuttgart.get_reconstructed_positions())
        print(stuttgart.get_ion_type())
```

## IVAS/APSuite (?) analysisset file reader

```python
prefix = f"{os.getcwd()}/../tests/data/analysisset/examples_with_provenance"
fnm = ["ZX21-T4.analysisset"]
```

```python
from ifes_apt_tc_data_modeling.analysisset.analysisset_reader import (
    ReadAnalysissetFileFormat,
)

for fpath in fnm:
    if fpath.lower().endswith(".analysisset"):
        print(fpath)
        analysisset = ReadAnalysissetFileFormat(f"{prefix}/{fpath}")
        # for m_ion in analysisset.analysisset["molecular_ions"]:
        #     m_ion.report()
        with h5py.File(f"{prefix}/{fpath}.nxs", "w") as h5w:
            h5w.create_group("/entry1")
            idx = 1
            for ion in analysisset.analysisset["molecular_ions"]:
                simple_hfive_file(h5w, idx, ion)
                idx += 1
```

## ATO

```python
prefix = f"{os.getcwd()}/../tests/data/ato/examples_without_provenance"
compare = False  # True for APSuite converted ATO to EPOS checks
fnm = [
    "Si.epos.v3.ATO",
    "Si.epos.v5.ATO",
]  # Si.epos example converted using tool from Cameca/APSuite from EPOS>ATO v3 and v5
if not compare:
    fnm = [
        "finfet_system.ato",
        "multilayer_system.ato",
        "R31_01853-v11_Fig2.ATO",
        "R31_03988-v05_Fig3.ATO",
        "R31_03992-v02_S2(a).ATO",
        "R31_03995-v02_S2(b).ATO",
        "R31_11378-v01.ato",
        "R31_11381-v02.ato",
        "R31_11553-v01.ato",
        "R31_11554-v01.ato",
        "R31_11556-v01.ato",
    ]
for fpath in fnm:
    print(os.path.getsize(f"{prefix}/{fpath}"))
```

```python
from ifes_apt_tc_data_modeling.ato.ato_reader import ReadAtoFileFormat

# the next line only to compare with ePOS
from ifes_apt_tc_data_modeling.epos.epos_reader import ReadEposFileFormat
```

```python
for fpath in fnm:
    print(f"{prefix}/{fpath}")
    ato = ReadAtoFileFormat(f"{prefix}/{fpath}")
    xyz = ato.get_reconstructed_positions()
    for idx, dim in enumerate(["x", "y", "z"]):
        print(
            f"{dim}: {np.min(xyz.magnitude[:, idx])}, {np.max(xyz.magnitude[:, idx])}"
        )
    mq = ato.get_mass_to_charge_state_ratio()
    print(f"m/q: {np.min(mq.magnitude)}, {np.max(mq.magnitude)}")

    # compare with original Si.epos
    if compare:
        epos = ReadEposFileFormat(f"{prefix}/Si.epos")
        print(epos.file_path)
        print(epos.file_size)
        xyz = epos.get_reconstructed_positions()
        for idx, dim in enumerate(["x", "y", "z"]):
            print(
                f"{dim}: {np.min(xyz.magnitude[:, idx])}, {np.max(xyz.magnitude[:, idx])}"
            )
        mq = epos.get_mass_to_charge_state_ratio()
        print(f"m/q: {np.min(mq.magnitude)}, {np.max(mq.magnitude)}")
```
